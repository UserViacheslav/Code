```python
import java.util.Random;
import java.util.Scanner;

public class Main {
    public static void main(String[] args) {
        Scanner scanner = new Scanner(System.in);
        Random random = new Random();

        int size = scanner.nextInt();
        int[] array = new int[size];

        System.out.print("[");
        for (int i = 0; i < size; i++) {
            array[i] = random.nextInt(11) - 5;
            System.out.print(array[i]);
            if (i < size - 1) {
                System.out.print(", ");
            }
        }
        System.out.println("]");

        int zeroCount = 0;
        int firstIndex = -1;
        int lastIndex = -1;

        for (int i = 0; i < array.length; i++) {
            if (array[i] == 0) {
                zeroCount++;
                if (firstIndex == -1) {
                    firstIndex = i;
                }
                lastIndex = i;
            }
        }

        if (zeroCount == 0) {
            System.out.println("Результат: Нульових елементів немає.");
        } else {
            System.out.println("Результат: Кількість нулів = " + zeroCount);
            System.out.println("Індекс першого нульового елемента: " + firstIndex);
            System.out.println("Індекс останнього нульового елемента: " + lastIndex);
        }

        scanner.close();
    }
}
```
