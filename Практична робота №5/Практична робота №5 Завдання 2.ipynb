```python
import java.util.Scanner;

public class Main {
    public static void main(String[] args) {
        double a = 0.1;
        double b = 1.4;
        double alpha = 0.02;
        double z = 0.003; 
        double k = 4.5;

        double x1 = Math.exp(Math.sqrt(1 - a * Math.sin(b)));
        double y1 = (a * k) / (4 * Math.log((alpha - z) / z));

        System.out.println("Завдання №1");
        System.out.println("Результат обчислень:");
        System.out.printf("x = %.5f\n", x1);
        System.out.printf("y = %.5f\n\n", y1);

        Scanner in = new Scanner(System.in);

        System.out.println("Завдання №2");
        System.out.print("Введіть a: ");
        double a2 = in.nextDouble();
        System.out.print("Введіть b: ");
        double b2 = in.nextDouble();
        System.out.print("Введіть значення Xmin: ");
        double xMin = in.nextDouble();
        System.out.print("Введіть значення Xmax: ");
        double xMax = in.nextDouble();
        System.out.print("Введіть значення dX: ");
        double dX = in.nextDouble();

        System.out.println("\nРезультати обчислень:");

        double currentX = xMin;
        while (currentX <= xMax + 0.00001) {
            double fx = 0;

            if (currentX < -2) {
                fx = Math.sin(currentX);
            } 
            if (currentX == -2) {
                fx = 2 * Math.exp(a2 * currentX) - 1;
            } 
            if (currentX > -2 && currentX < 8) {
                fx = 1 / ((b2 * currentX + 2 * a2) * (b2 * currentX + 2 * a2));
            }

            System.out.printf("f(%.5f) = %.5f\n", currentX, fx);
            currentX = currentX + dX;
        }

        in.close();
    }
}
```
