```python
public class Main {
    public static void main(String[] args) {
        double a = 0.1;
        double b = 1.4;
        double alpha = 0.02;
        double z = 0.003; 
        double k = 4.5;

        double x = Math.exp(Math.sqrt(1 - a * Math.sin(b)));
        double y = (a * k) / (4 * Math.log((alpha - z) / z));

        System.out.println("Завдання №1");
        System.out.println("Результат обчислень:");
        System.out.printf("x = %.5f\n", x);
        System.out.printf("y = %.5f\n", y);
    }
}
```
